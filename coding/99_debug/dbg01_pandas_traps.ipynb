### 题 3 · 分组统计写成了聚合

想给每一行都填上「它所在组的中位数」。

跑一遍看看：`按组中位数` 整列是 `NaN`。

```python

import pandas as pd

d = pd.DataFrame({"组": ["x", "x", "y", "y"], "值": [1.0, 3.0, 10.0, 20.0]})

d["按组中位数"] = d.groupby("组")["值"].median()

print(d)
```
